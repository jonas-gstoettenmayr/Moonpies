```python
# Setup environment
from pathlib import Path
from dotenv import load_dotenv

# Load environment variables from meteo-api.env
load_dotenv(Path.cwd() / 'meteo-api.env')

# Add src to path so we can import moonpies
import sys
sys.path.insert(0, str(Path.cwd() / 'src'))
```

```python
# Import the meteomatics client
from moonpies import get_weather
print('✓ Successfully imported get_weather from moonpies package')
```

```python
from datetime import datetime, timezone

# Fetch weather for Graz, Austria
weather = get_weather(
    latitude=47.07,
    longitude=15.44,
    valid_time=datetime.now(timezone.utc),
)

print(weather)
```

```python
# Parse and display weather data
if 'data' in weather:
    for data_point in weather['data']:
        param = data_point.get('parameter', 'unknown')
        for coord in data_point.get('coordinates', []):
            for date_info in coord.get('dates', []):
                print(f"{param}: {date_info.get('value')} at {date_info.get('date')}")
```
